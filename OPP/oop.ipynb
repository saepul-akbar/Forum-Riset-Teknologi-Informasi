Object Oriented Programming

OOP adalah suatu metode pemrograman yang fokus atau berorientasi pada objek.
OOP membantu kita mengelola kompleksitas dengan membungkusnya menjedi sebuah objek.

OOP dibagi menjadi 2 yaitu class dan object

```python
class Mobil: # Mobil adalah nama class
    #methods
    def start(self): 
        print("Mobil dinyalakan")
    def stop(self):
        print("Mobil dimatikan")

mobil = Mobil() #mobil adalah object
mobil.start()
mobil.stop()
```

```python
class Mobil: # Mobil adalah nama class
    merek = "Toyota"
    warna = "Merah"
    tahun = 2020

mobil = Mobil() #mobil adalah object
print(mobil.warna)
print(mobil.merek)
print(mobil.tahun)
```

Prinsip oop

```python
from abc import ABC, abstractmethod

# Abstraction
class Animal(ABC): # abstract class
    def __init__(self, name):
        self._name = name # encapsulation (protected attribute)
    
    @abstractmethod
    def speak(self): # abstract method
        pass

# Inheritance
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)

    # Polymorphism
    def speak(self):
        return f"{self._name} says: Woof!"

class Cat(Animal):
    def __init__(self, name):
        super().__init__(name)
    
    # Polymorphism
    def speak(self):
        return f"{self._name} says: Meow!"

# Main
dog = Dog("Buddy")
cat = Cat("Mimi")

print(dog.speak())
print(cat.speak())
```

```python
# --- TANTANGAN UNTUK SISWA: GARASI SEDERHANA ---

from abc import ABC, abstractmethod

# TUGAS 1: ABSTRACTION (Sudah Selesai)
# Ini adalah 'kontrak' yang harus dipatuhi semua Kendaraan.
class Kendaraan(ABC):

    @abstractmethod
    def bergerak(self):
        pass

class Mobil(Kendaraan):
    def __init__(self, nama):
        self.nama = nama
        self.__bensin = 0
    
    def isi_bensin(self, jumlah):
        self.__bensin = self.__bensin + jumlah

        print(f"{self.nama} di isi bensin {jumlah} L")
    
    def bergerak(self):
        if self.__bensin > 0:
            print("Vroom Vroom!")
            self.__bensin -= 1
        else:
            print("Bensin habis!")

class Sepeda(Kendaraan):
    def __init__(self, nama):
        self.nama = nama
    
    def bergerak(self):
        print(f"Sepeda {self.nama} bergerak...Sssshhh!")

# --- UJI COBA KODE (Jangan diubah) ---
mobil_A = Mobil("Sedan")
sepeda_B = Sepeda("BMX")
garasi = [mobil_A, sepeda_B]

print("--- Cek Garasi (Polymorphism) ---")
for kendaraan in garasi:
    kendaraan.bergerak()

print("\n--- Mengisi Bensin Mobil ---")
mobil_A.isi_bensin(10)

print("\n--- Coba Gerakkan Mobil Lagi ---")
mobil_A.bergerak()


```
