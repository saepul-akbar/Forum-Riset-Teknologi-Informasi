Tipe Data

```python
x = 10
y = 3.14
z = "Hello" #string
a = True #boolean
b = [1,2,3] #list
c = (4, 5, 6) #tuple
d = {"nama": "Saepul", "umur": 19} #dictionary
e = {7, 8, 9} #set

print(type(x))
print(type(y))
print(type(z))
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

Mengecek tipe data variable python

```python
umur = "12"
print(type(umur))
```

Memanggil index dalam list

```python
b = [1,2,3]
print(b[2])
```

Memanggil spesifik index dalam nested list

```python
data = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(data[1][2])  # Output: 6
```

Mengambil data dari dictionary

```python
data_dict = {"nama": "Saepul", "umur": 19}
print(data_dict["nama"])  # Output: Saepul
print(data_dict["umur"])  # Output: 19
```
